# Exploratory Data Analysis - Healthcare Fraud Detection
Generates visualisations for fraud vs non-fraud, distributions, correlations, trends.
Savings to reports/.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from pathlib import Path

master = pd.read_csv("../reports/master_provider_features.csv")
master.head()
```

```python
# Class distribution
print(master["PotentialFraud"].value_counts())
print(master["PotentialFraud"].value_counts(normalize=True))

plt.figure(figsize=(6,4))
sns.countplot(data=master, x="PotentialFraud")
plt.title("Class Distribution: Fraud vs Non-Fraud")
plt.savefig("../reports/eda_class_distribution.png", dpi=200)
plt.show()
```

```python
# Reimbursement distributions
fig, axes = plt.subplots(2,2, figsize=(14,10))
sns.histplot(data=master, x="TotalReimbursement", hue="PotentialFraud", bins=50, ax=axes[0,0])
axes[0,0].set_title("Total Reimbursement")
sns.boxplot(data=master, x="PotentialFraud", y="AverageReimbursement", ax=axes[0,1])
axes[0,1].set_title("Avg Reimbursement by Fraud")
sns.histplot(data=master, x="TotalClaims", hue="PotentialFraud", bins=50, ax=axes[1,0])
axes[1,0].set_title("Total Claims")
sns.boxplot(data=master, x="PotentialFraud", y="UniqueBeneficiaries", ax=axes[1,1])
axes[1,1].set_title("Unique Beneficiaries by Fraud")
plt.tight_layout()
plt.savefig("../reports/eda_reimbursement.png", dpi=200)
plt.show()
```

```python
# Correlation heatmap
numeric = master.select_dtypes(include=[np.number])
corr = numeric.corr()
plt.figure(figsize=(14,12))
sns.heatmap(corr, cmap="coolwarm", center=0, annot=False)
plt.title("Correlation Heatmap - Provider Features")
plt.tight_layout()
plt.savefig("../reports/eda_correlation.png", dpi=200)
plt.show()
```
